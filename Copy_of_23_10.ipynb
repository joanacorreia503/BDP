**Recursion - Programming Technique**

```python
def f(n):
  if n == 0:
    return 1
  else:
    return n * f(n - 1)

f(5)
```

```python
def C(n, p):
  if n == p or p == 0:
    return 1
  else:
    return C(n - 1, p - 1) + C(n - 1, p)

C(5,3)
```

```python
def fib(n):
  if n < 3:
    return 1
  else:
    return fib(n - 1) + fib(n - 2)

fib(10)
```

```python
def fib(n, counters):
  counters[n] += 1
  if n < 3:
    return 1
  else:
    return fib(n - 1, counters) + fib(n - 2, counters)

w = [0] * 21
fib(20, w)
```

```python
def fib(n, mem):
  if mem[n] is None:
    if n < 3:
      mem[n] = 1
    else:
      mem[n] = fib(n - 1, mem) + fib(n - 2, mem)

  return mem[n]

M = [None] * 101
fib(100, M)
```

```python
def C(n, p, D):
  key = str(n) + "," + str(p)
  if key in D:
    return D[key]
  else:
    if n == p or p == 0:
      D[key] = 1
    else:
      D[key] = C(n - 1, p - 1, D) + C(n - 1, p, D)

  return D[key]

D = {}

C(50, 32, D)
```

```python
#tower of hanoi game
def H(n, x, y, z):
  if n > 0:
    H(n - 1, x, z, y)
    print("Move from ", x, "to ", z)
    H(n - 1, y, x, z)
  else:
    return []

H (3, 1, 2, 3)
```
