```python
def gcd(A, B):
  while B>0:
    A,B = B,A%B
  return A
```

```python
gcd(10,2)
```

```python
def NewtonSQR (N, epsilon):
  x = N/2
  d = epsilon
  while d >= epsilon:
    xprev = x
    x = x - (x**2 - N) / (2*x)
    d = abs(x - xprev)

  return x
```

```python
NewtonSQR(2, 0.00001)
```
